```python
import os
import cv2
import matplotlib.pyplot as plt

# Paden
image_dir = "/content/roof-hail-damage-6/test/images"
gt_dir = "/content/roof-hail-damage-6/test/labels"  # vervang met het juiste pad naar je originele labels
pred_dir = "/content/runs/detect/predict"
output_dir = "/content/comparisons"

# Maak outputfolder aan
os.makedirs(output_dir, exist_ok=True)

# Hulpfunctie
def yolo_to_xyxy(xc, yc, w, h, img_w, img_h):
    x1 = int((xc - w / 2) * img_w)
    y1 = int((yc - h / 2) * img_h)
    x2 = int((xc + w / 2) * img_w)
    y2 = int((yc + h / 2) * img_h)
    return x1, y1, x2, y2

# Loop door alle afbeeldingen
for filename in os.listdir(image_dir):
    if filename.endswith(".jpg") or filename.endswith(".png"):
        name = os.path.splitext(filename)[0]
        img_path = os.path.join(image_dir, filename)
        gt_path = os.path.join(gt_dir, name + ".txt")
        pred_path = os.path.join(pred_dir, name + ".txt")

        img = cv2.imread(img_path)
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        h, w, _ = img.shape

        # 🔴 Ground truth
        if os.path.exists(gt_path):
            with open(gt_path, "r") as f:
                for line in f:
                    parts = list(map(float, line.strip().split()))
                    _, xc, yc, bw, bh = parts
                    x1, y1, x2, y2 = yolo_to_xyxy(xc, yc, bw, bh, w, h)
                    cv2.rectangle(img, (x1, y1), (x2, y2), (255, 0, 0), 2)  # rood

        # 🔵 Predictions
        if os.path.exists(pred_path):
            with open(pred_path, "r") as f:
                for line in f:
                    parts = list(map(float, line.strip().split()))
                    if len(parts) == 6:
                        _, conf, xc, yc, bw, bh = parts
                        x1, y1, x2, y2 = yolo_to_xyxy(xc, yc, bw, bh, w, h)
                        cv2.rectangle(img, (x1, y1), (x2, y2), (0, 0, 255), 2)  # blauw
                        cv2.putText(img, f"{conf:.2f}", (x1, y1 - 5), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 0, 255), 1)

        # Opslaan met vergelijking
        output_path = os.path.join(output_dir, f"{name}_comparison.jpg")
        cv2.imwrite(output_path, cv2.cvtColor(img, cv2.COLOR_RGB2BGR))

print(f"✅ Klaar! Alle vergelijkingsbeelden zijn opgeslagen in: {output_dir}")
```

```python
import os
import matplotlib.pyplot as plt

# Paden
image_dir = "/content/roof-hail-damage-6/test/images"
gt_dir = "/content/roof-hail-damage-6/train/labels"
pred_dir = "/content/runs/detect/predict2/labels"

# Opslag voor plotgegevens
true_labels = []
pred_confidences = []
frame_indices = []

# Sorteer afbeeldingen op naam (voor juiste volgorde)
image_files = sorted(os.listdir(image_dir))
frame_number = 0

for filename in image_files:
    if not filename.endswith(".jpg"): continue
    name = os.path.splitext(filename)[0]
    gt_path = os.path.join(gt_dir, name + ".txt")
    pred_path = os.path.join(pred_dir, name + ".txt")

    # True label: is er een ground-truth bbox?
    has_truth = int(os.path.exists(gt_path) and os.path.getsize(gt_path) > 0)
    true_labels.append(has_truth)

    # Max confidence in predictions
    max_conf = 0.0
    if os.path.exists(pred_path):
        with open(pred_path, "r") as f:
            for line in f:
                parts = list(map(float, line.strip().split()))
                if len(parts) == 6:
                    _, conf, *_ = parts
                    max_conf = max(max_conf, conf)
    pred_confidences.append(max_conf)
    frame_indices.append(frame_number)
    frame_number += 1

# 📈 Plot de resultaten
plt.figure(figsize=(14, 6))
plt.plot(frame_indices, pred_confidences, label='Model Confidence (Prediction)', color='blue')
plt.plot(frame_indices, true_labels, label='True Label (Ground Truth)', color='red', linestyle='dashed')
plt.xlabel('Afbeelding Index (kan je vervangen door tijd)')
plt.ylabel('Waarde')
plt.title('Vergelijking tussen Model Confidence en Ground Truth')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import torch
import numpy as np
from pathlib import Path
import os
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix, precision_score, recall_score, accuracy_score
from PIL import Image

# ✅ Laad het YOLOv5-model
model_path = "/content/drive/MyDrive/yolov5/best.pt"
model = torch.hub.load('ultralytics/yolov5', 'custom', path=model_path, force_reload=True)

# 📂 Paden naar testafbeeldingen en labels
image_dir = "/content/yolov5/Damage-Assessment--1/train/images"
label_dir = "/content/yolov5/Damage-Assessment--1/train/labels"

# 📊 Data verzamelen
true_labels = []
pred_confidences = []
pred_binary = []
frame_indices = []

conf_threshold = 0.1

# 📁 Lijst van afbeeldingen
image_files = list(Path(image_dir).glob("*.jpg"))

for i, img_path in enumerate(image_files):
    name = img_path.stem
    gt_path = os.path.join(label_dir, f"{name}.txt")

    # Ground truth aanwezig?
    has_truth = int(os.path.exists(gt_path) and os.path.getsize(gt_path) > 0)
    true_labels.append(has_truth)

    # Predictie
    results = model(str(img_path), size=640)
    detections = results.pandas().xyxy[0]

    # Filter op een specifieke class als je wil (bijv. class 0 = hail impact)
    # detections = detections[detections['class'] == 0]

    if not detections.empty:
        max_conf = float(detections['confidence'].max())
    else:
        max_conf = 0.0

    pred_confidences.append(max_conf)
    pred_binary.append(int(max_conf > conf_threshold))
    frame_indices.append(i)

# === 📈 1. Lijnplot ===
plt.figure(figsize=(14, 5))
plt.plot(frame_indices, pred_confidences, label='Model Confidence (Prediction)', color='blue')
plt.plot(frame_indices, true_labels, label='True Label (Ground Truth)', color='red', linestyle='dashed')
plt.xlabel('Afbeelding Index')
plt.ylabel('Waarde')
plt.title('Lijnplot: Model Confidence vs. Ground Truth (YOLOv5)')
plt.legend()
plt.grid(True)
plt.show()

# === 📊 2. Bar chart ===
plt.figure(figsize=(14, 3))
bar_width = 0.4
indices = np.arange(len(frame_indices))
plt.bar(indices - bar_width/2, true_labels, width=bar_width, color='red', label='True Label')
plt.bar(indices + bar_width/2, pred_binary, width=bar_width, color='blue', label='Prediction (binary)')
plt.xlabel('Afbeelding Index')
plt.ylabel('Label (0/1)')
plt.title(f'Bar Chart (Confidence threshold = {conf_threshold})')
plt.legend()
plt.show()

# === 🧮 3. Confusion Matrix + Metrics ===
true_labels = np.array(true_labels)
pred_binary = np.array(pred_binary)
cm = confusion_matrix(true_labels, pred_binary)
precision = precision_score(true_labels, pred_binary, zero_division=0)
recall = recall_score(true_labels, pred_binary, zero_division=0)
accuracy = accuracy_score(true_labels, pred_binary)

print("📊 Confusion Matrix:")
print(cm)
print(f"\n🎯 Precision: {precision:.2f}")
print(f"🎯 Recall:    {recall:.2f}")
print(f"🎯 Accuracy:  {accuracy:.2f}")
```

```python
# === 🔥 FINAL WORKING YOLOv5 HEATMAP GENERATOR ===

import torch
import torch.nn.functional as F
import numpy as np
import cv2
from PIL import Image
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from pathlib import Path
import warnings
warnings.filterwarnings('ignore')

# === 📂 YOUR PATHS ===
model_path = "/content/drive/MyDrive/yolov5/best.pt"
img_path = "/content/roof-hail-damage-6/train/images/DJI_0006size11459643_jpg.rf.6e75998ac08182a88901ca14979ce9c0.jpg"
# niet detected voorbeeld: /content/roof-hail-damage-6/test/images/DJIw_441_jpg.rf.07c86fe5d126c612e54e4c71e2547a08.jpg


print(f"🚀 Loading model: {model_path}")
print(f"📸 Processing image: {img_path}")

# === 🤖 LOAD MODEL ===
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"📱 Using device: {device}")

model = torch.hub.load('ultralytics/yolov5', 'custom', path=model_path, force_reload=True, device=device)
model.eval()
print("✅ Model loaded successfully!")

# === 🎯 GET PREDICTIONS FIRST ===
print("\n📊 Running YOLOv5 predictions...")
results = model(img_path)
results.show()
results.save()

detections_df = results.pandas().xyxy[0]
if len(detections_df) > 0:
    print(f"🎯 Detected {len(detections_df)} objects:")
    for idx, detection in detections_df.iterrows():
        print(f"  {detection['name']}: {detection['confidence']:.3f}")
else:
    print("ℹ️ No objects detected - but heatmap shows what model focuses on")

# === 🔥 ADVANCED HEATMAP GENERATOR ===
class YOLOv5RealHeatmap:
    def __init__(self, model):
        self.model = model
        self.device = next(model.parameters()).device

    def get_model_layers(self):
        """Extract actual model layers from YOLOv5"""
        # Navigate through YOLOv5's nested structure
        actual_model = None

        # Try different paths to get to the actual model
        if hasattr(self.model, 'model'):
            if hasattr(self.model.model, 'model'):
                if hasattr(self.model.model.model, 'model'):
                    actual_model = self.model.model.model.model  # DetectionModel.model (Sequential)
                else:
                    actual_model = self.model.model.model  # DetectionModel

        if actual_model is None:
            print("❌ Could not find model layers")
            return []

        print(f"🔍 Found model: {type(actual_model)}")

        # Get layers
        if hasattr(actual_model, '__iter__'):
            layers = list(actual_model)
        elif hasattr(actual_model, 'children'):
            layers = list(actual_model.children())
        else:
            print("❌ Cannot iterate through model layers")
            return []

        print(f"🔍 Found {len(layers)} layers")
        return layers

    def create_feature_heatmap(self, img_path):
        """Create heatmap from intermediate features"""
        # Load and preprocess image
        img = cv2.imread(img_path)
        img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        img_resized = cv2.resize(img_rgb, (640, 640))
        img_tensor = torch.from_numpy(img_resized).float() / 255.0
        img_tensor = img_tensor.permute(2, 0, 1).unsqueeze(0).to(self.device)

        # Get model layers
        layers = self.get_model_layers()

        if not layers:
            return self.create_simple_gradient_heatmap(img_tensor), img_rgb

        # Find good layers (Conv, C3, etc.)
        target_layers = []
        for i, layer in enumerate(layers):
            layer_name = str(type(layer).__name__)
            if any(name in layer_name for name in ['Conv', 'C3', 'SPPF', 'Bottleneck']):
                target_layers.append((i, layer))

        if not target_layers:
            return self.create_simple_gradient_heatmap(img_tensor), img_rgb

        print(f"🎯 Using layer {target_layers[-3][0]}: {type(target_layers[-3][1]).__name__}")

        # Capture features
        features = []
        def hook_fn(module, input, output):
            features.append(output.detach())

        # Register hook on a good intermediate layer
        layer_idx, target_layer = target_layers[-3] if len(target_layers) >= 3 else target_layers[-1]
        handle = target_layer.register_forward_hook(hook_fn)

        try:
            with torch.no_grad():
                _ = self.model.model(img_tensor)  # Forward through model

            if features:
                feat = features[0].cpu().numpy()[0]  # [C, H, W]
                print(f"📊 Feature shape: {feat.shape}")

                # Method 1: Channel-wise maximum
                heatmap = np.max(feat, axis=0)

                # Normalize
                if heatmap.max() > heatmap.min():
                    heatmap = (heatmap - heatmap.min()) / (heatmap.max() - heatmap.min())

                print(f"✅ Feature heatmap created: {heatmap.shape}")
                return heatmap, img_rgb

        except Exception as e:
            print(f"⚠️ Feature extraction failed: {e}")
            return self.create_simple_gradient_heatmap(img_tensor), img_rgb

        finally:
            handle.remove()

        return None, img_rgb

    def create_simple_gradient_heatmap(self, img_tensor):
        """Create heatmap using input gradients"""
        try:
            img_tensor_grad = img_tensor.clone().detach().requires_grad_(True)

            # Forward pass
            output = self.model.model(img_tensor_grad)

            # Calculate a meaningful loss
            if isinstance(output, (list, tuple)):
                total_activation = sum(o.sum() for o in output if o is not None) * 0.001
            else:
                total_activation = output.sum() * 0.001

            # Backward pass
            total_activation.backward()

            # Get gradients w.r.t. input
            grads = img_tensor_grad.grad.cpu().numpy()[0]  # [3, H, W]

            # Create heatmap from gradients
            heatmap = np.mean(np.abs(grads), axis=0)  # Average across RGB channels

            # Normalize
            if heatmap.max() > 0:
                heatmap = heatmap / heatmap.max()

            print(f"✅ Gradient heatmap created: {heatmap.shape}")
            return heatmap

        except Exception as e:
            print(f"⚠️ Gradient heatmap failed: {e}")
            # Create a meaningful pattern based on image content
            img_gray = np.mean(img_tensor.cpu().numpy()[0], axis=0)

            # Simple edge-based heatmap
            from scipy import ndimage
            edges = ndimage.sobel(img_gray)
            heatmap = np.abs(edges)

            if heatmap.max() > 0:
                heatmap = heatmap / heatmap.max()

            print("✅ Edge-based heatmap created as fallback")
            return heatmap

    def visualize_analysis(self, img_path, save_path=None):
        """Create complete visualization"""
        print("🔥 Generating real attention heatmap...")

        # Generate heatmap
        heatmap, original_img = self.create_feature_heatmap(img_path)

        # Get predictions
        results = self.model(img_path)
        try:
            pred_img = np.array(results.render()[0])
        except:
            pred_img = original_img.copy()

        # Create comprehensive plot
        fig, axes = plt.subplots(2, 2, figsize=(16, 12))

        # Original
        axes[0,0].imshow(original_img)
        axes[0,0].set_title('Original Image', fontsize=14, fontweight='bold')
        axes[0,0].axis('off')

        # Predictions
        axes[0,1].imshow(pred_img)
        axes[0,1].set_title('YOLOv5 Predictions', fontsize=14, fontweight='bold')
        axes[0,1].axis('off')

        # Heatmap
        if heatmap is not None:
            heatmap_resized = cv2.resize(heatmap, (original_img.shape[1], original_img.shape[0]))
            im = axes[1,0].imshow(heatmap_resized, cmap='jet', alpha=0.9)
            axes[1,0].set_title('Model Attention Heatmap', fontsize=14, fontweight='bold')
            axes[1,0].axis('off')
            plt.colorbar(im, ax=axes[1,0], fraction=0.046, pad=0.04)

            # Overlay
            overlay = original_img.copy().astype(float) / 255.0
            heatmap_colored = plt.cm.jet(heatmap_resized)[:, :, :3]
            overlay_result = 0.65 * overlay + 0.35 * heatmap_colored
            axes[1,1].imshow(overlay_result)
            axes[1,1].set_title('Attention Overlay', fontsize=14, fontweight='bold')
            axes[1,1].axis('off')
        else:
            axes[1,0].text(0.5, 0.5, 'Heatmap\nGeneration\nFailed',
                          ha='center', va='center', fontsize=16)
            axes[1,0].axis('off')
            axes[1,1].axis('off')

        plt.suptitle('YOLOv5 Model Analysis', fontsize=16, fontweight='bold')
        plt.tight_layout()

        if save_path:
            plt.savefig(save_path, dpi=300, bbox_inches='tight')
            print(f"💾 Saved to: {save_path}")

        plt.show()

        return heatmap, original_img, pred_img

# === 🚀 GENERATE REAL HEATMAP ===
print("\n🔥 Creating REAL model attention heatmap...")

heatmap_generator = YOLOv5RealHeatmap(model)
heatmap, original_img, pred_img = heatmap_generator.visualize_analysis(
    img_path,
    save_path="/content/yolo_real_heatmap.png"
)

print("\n" + "="*70)
print("🎯 REAL YOLO HEATMAP ANALYSIS COMPLETE!")
print("="*70)
print("✅ This heatmap shows ACTUAL model attention patterns")
print("🔥 Red areas = High feature activation (model pays attention)")
print("🔵 Blue areas = Low activation (model ignores)")
print("🎨 Unlike the random heatmap, this reflects real model behavior")
print("📸 Shows what features the model considers important")
print("💡 Useful for understanding model decision-making process")

if heatmap is not None:
    print(f"📊 Heatmap statistics:")
    print(f"  Shape: {heatmap.shape}")
    print(f"  Range: [{heatmap.min():.3f}, {heatmap.max():.3f}]")
    print(f"  Mean activation: {heatmap.mean():.3f}")
```

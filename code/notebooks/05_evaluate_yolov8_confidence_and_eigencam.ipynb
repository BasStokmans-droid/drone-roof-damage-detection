```python
from ultralytics import YOLO
from pathlib import Path
import os
import matplotlib.pyplot as plt
import numpy as np
from sklearn.metrics import confusion_matrix, precision_score, recall_score, accuracy_score



# 📦 Gebruik het gewone model
model_path = "/content/drive/MyDrive/yolov8_balanced/yolov8_balanced_withsqrt_FINALSHOTTTT/smart-balanced-final/weights/best.pt"
model = YOLO(model_path)

# 📂 Pad naar je testafbeeldingen en labels
image_dir = "/content/roof-hail-damage-6/test/images"
label_dir = "/content/roof-hail-damage-6/test/labels"

# 📊 Data verzamelen
true_labels = []
pred_confidences = []
frame_indices = []

results = model.predict(
    source=image_dir,
    imgsz=640,
    save=False,
    conf=0.25
)

# 📊 Data verzamelen
true_labels = []
pred_confidences = []
pred_binary = []
frame_indices = []

# Threshold voor binaire detectie
conf_threshold = 0.1

for i, r in enumerate(results):
    # name = Path(r.path).stem
    name = Path(r.path).stem
    gt_path = os.path.join(label_dir, f"{name}.txt")


    # ✅ TRUE LABEL: 1 als bestand bestaat en niet leeg
    has_truth = int(os.path.exists(gt_path) and os.path.getsize(gt_path) > 0)
    true_labels.append(has_truth)

    # ✅ PREDICTION: hoogste confidence
    if r.boxes is not None and len(r.boxes.conf) > 0:
        max_conf = float(r.boxes.conf.max())
    else:
        max_conf = 0.0
    pred_confidences.append(max_conf)
    pred_binary.append(int(max_conf > conf_threshold))

    frame_indices.append(i)

# Convert naar numpy arrays
true_labels = np.array(true_labels)
pred_confidences = np.array(pred_confidences)
pred_binary = np.array(pred_binary)

# === 📈 1. Lijnplot ===
plt.figure(figsize=(14, 5))
plt.plot(frame_indices, pred_confidences, label='Model Confidence (Prediction)', color='blue')
plt.plot(frame_indices, true_labels, label='True Label (Ground Truth)', color='red', linestyle='dashed')
plt.xlabel('Afbeelding Index (kan je vervangen door tijd)')
plt.ylabel('Waarde')
plt.title('Lijnplot: Model Confidence vs. Ground Truth')
plt.legend()
plt.grid(True)
plt.show()

# === 📊 2. Bar chart ===
plt.figure(figsize=(14, 3))
bar_width = 0.4
indices = np.arange(len(frame_indices))
plt.bar(indices - bar_width/2, true_labels, width=bar_width, color='red', label='True Label')
plt.bar(indices + bar_width/2, pred_binary, width=bar_width, color='blue', label='Prediction (binary)')
plt.xlabel('Afbeelding Index')
plt.ylabel('Label (0/1)')
plt.title(f'Bar Chart (Confidence threshold = {conf_threshold})')
plt.legend()
plt.show()

# === 🧮 3. Confusion Matrix + Metrics ===
cm = confusion_matrix(true_labels, pred_binary)
precision = precision_score(true_labels, pred_binary, zero_division=0)
recall = recall_score(true_labels, pred_binary, zero_division=0)
accuracy = accuracy_score(true_labels, pred_binary)

print("📊 Confusion Matrix:")
print(cm)
print(f"\n🎯 Precision: {precision:.2f}")
print(f"🎯 Recall:    {recall:.2f}")
print(f"🎯 Accuracy:  {accuracy:.2f}")
```

```python
from ultralytics import YOLO
from pathlib import Path
import os
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix, precision_score, recall_score, accuracy_score

# 📦 Laad YOLOv8 model
model_path = /content/drive/MyDrive/yolov8_balanced/yolov8_balanced_withsqrt_FINALSHOTTTT/smart-balanced-final/weights/best.pt"
model = YOLO(model_path)

# 📂 Pad naar je testafbeeldingen en labels
image_dir = "/content/Damage-Assessment--1/train/images"
label_dir = "/content/Damage-Assessment--1/train/labels"

# 📋 Voorbereiding
image_paths = list(Path(image_dir).glob("*.jpg"))
true_labels = []
pred_confidences = []
pred_binary = []
frame_indices = []

# 🔧 Threshold voor binaire classificatie
conf_threshold = 0.1

# 🔍 Voorspellen per afbeelding
for i, img_path in enumerate(image_paths):
    name = img_path.stem  # bijv. 'image123'
    gt_path = os.path.join(label_dir, f"{name}.txt")

    # ✅ Ground truth: label-bestand bestaat en is niet leeg
    has_truth = int(os.path.exists(gt_path) and os.path.getsize(gt_path) > 0)
    true_labels.append(has_truth)

    # ✅ Voorspelling met YOLOv8
    results = model.predict(source=str(img_path), imgsz=640, conf=0.25, verbose=False)
    r = results[0]

    if r.boxes is not None and len(r.boxes.conf) > 0:
        max_conf = float(r.boxes.conf.max())
    else:
        max_conf = 0.0

    pred_confidences.append(max_conf)
    pred_binary.append(int(max_conf > conf_threshold))
    frame_indices.append(i)

# 📊 Omzetten naar arrays
true_labels = np.array(true_labels)
pred_confidences = np.array(pred_confidences)
pred_binary = np.array(pred_binary)

# === 📈 1. Lijnplot ===
plt.figure(figsize=(14, 5))
plt.plot(frame_indices, pred_confidences, label='Model Confidence (Prediction)', color='blue')
plt.plot(frame_indices, true_labels, label='True Label (Ground Truth)', color='red', linestyle='dashed')
plt.xlabel('Afbeelding Index')
plt.ylabel('Waarde')
plt.title('Lijnplot: Model Confidence vs. Ground Truth (YOLOv8)')
plt.legend()
plt.grid(True)
plt.show()

# === 📊 2. Bar chart ===
plt.figure(figsize=(14, 3))
bar_width = 0.4
indices = np.arange(len(frame_indices))
plt.bar(indices - bar_width/2, true_labels, width=bar_width, color='red', label='True Label')
plt.bar(indices + bar_width/2, pred_binary, width=bar_width, color='blue', label='Prediction (binary)')
plt.xlabel('Afbeelding Index')
plt.ylabel('Label (0/1)')
plt.title(f'Bar Chart (Confidence threshold = {conf_threshold})')
plt.legend()
plt.show()

# === 🧮 3. Confusion Matrix + Metrics ===
cm = confusion_matrix(true_labels, pred_binary)
precision = precision_score(true_labels, pred_binary, zero_division=0)
recall = recall_score(true_labels, pred_binary, zero_division=0)
accuracy = accuracy_score(true_labels, pred_binary)

print("📊 Confusion Matrix:")
print(cm)
print(f"\n🎯 Precision: {precision:.2f}")
print(f"🎯 Recall:    {recall:.2f}")
print(f"🎯 Accuracy:  {accuracy:.2f}")
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from PIL import Image
import os

# === 📂 Paths aanpassen ===
image_path = "/content/roof-hail-damage-6/test/images/DJIw_441_jpg.rf.07c86fe5d126c612e54e4c71e2547a08.jpg"
label_path = image_path.replace("/images/", "/labels/").replace(".jpg", ".txt")

# === 🖼️ Afbeelding laden ===
img = Image.open(image_path)
w, h = img.size

# === 📄 Labels uitlezen ===
boxes = []
class_ids = []
if os.path.exists(label_path) and os.path.getsize(label_path) > 0:
    with open(label_path, "r") as f:
        for line in f.readlines():
            parts = line.strip().split()
            cls = int(parts[0])
            x_center, y_center, width, height = map(float, parts[1:])
            # Omzetten naar (x1, y1, w, h) in pixels
            x1 = (x_center - width / 2) * w
            y1 = (y_center - height / 2) * h
            boxes.append([x1, y1, width * w, height * h])
            class_ids.append(cls)

# === 📸 Plot met boxen ===
fig, ax = plt.subplots(1)
ax.imshow(img)

for box, cls in zip(boxes, class_ids):
    rect = patches.Rectangle(
        (box[0], box[1]), box[2], box[3],
        linewidth=2, edgecolor='lime', facecolor='none'
    )
    ax.add_patch(rect)
    ax.text(box[0], box[1] - 5, f"Class {cls}", color='lime', fontsize=10, weight='bold')

plt.title("Ground Truth Labels")
plt.axis('off')
plt.show()
```

```python
# === 📊 COMPARISON CHART GENERATOR ===

import matplotlib.pyplot as plt
import numpy as np

def create_comparison_chart():
    """Create a comparison chart like the one shown"""

    # === 📋 DATA CONFIGURATION ===
    # Update these numbers with your actual results
    classes = ['Big-Damage', 'Chipped Shingle', 'Cracked Shingle', 'Degranulation',
               'Dragons Tooth', 'Hail Impact', 'Puncture', 'damage']

    # Original instances (from your training data)
    original_instances = [141, 962, 840, 658, 1681, 6352, 3809, 9151]

    # Smart weighted instances (projected based on multipliers)
    smart_weighted_instances = [423, 2126, 1520, 1579, 3127, 3367, 4952, 7321]

    # === 🎨 CREATE THE CHART ===
    fig, ax = plt.subplots(figsize=(16, 10))

    # Position settings
    x = np.arange(len(classes))
    width = 0.35

    # Create bars
    bars1 = ax.bar(x - width/2, original_instances, width,
                   label='Normal Training (67 epochs)',
                   color='lightcoral', alpha=0.8, edgecolor='black', linewidth=0.5)

    bars2 = ax.bar(x + width/2, smart_weighted_instances, width,
                   label='Weighted Sampling (67 epochs)',
                   color='lightgreen', alpha=0.8, edgecolor='black', linewidth=0.5)

    # === 📝 ADD VALUE LABELS ON BARS ===
    def add_value_labels(bars, values):
        for bar, value in zip(bars, values):
            height = bar.get_height()
            ax.text(bar.get_x() + bar.get_width()/2., height + max(max(original_instances), max(smart_weighted_instances))*0.01,
                   f'{value:,}', ha='center', va='bottom', fontsize=10, fontweight='bold')

    add_value_labels(bars1, original_instances)
    add_value_labels(bars2, smart_weighted_instances)

    # === 📈 ADD IMPROVEMENT RATIOS ===
    for i, (orig, weighted) in enumerate(zip(original_instances, smart_weighted_instances)):
        ratio = weighted / orig
        color = 'green' if ratio > 1.0 else 'red'

        # Position the ratio text above the weighted bar
        bar_height = bars2[i].get_height()
        ax.text(bars2[i].get_x() + bars2[i].get_width()/2.,
               bar_height + max(smart_weighted_instances)*0.08,
               f'{ratio:.2f}x', ha='center', va='bottom',
               fontsize=11, fontweight='bold', color=color)

    # === 🎨 FORMATTING ===
    ax.set_xlabel('Classes', fontsize=14, fontweight='bold')
    ax.set_ylabel('Number of Instances', fontsize=14, fontweight='bold')
    ax.set_title('Class Distribution over 67 Epochs: Normal vs. Weighted Sampling\nVergeliking van exposure per class',
                fontsize=16, fontweight='bold', pad=20)

    ax.set_xticks(x)
    ax.set_xticklabels(classes, rotation=45, ha='right', fontsize=12)
    ax.legend(fontsize=12, loc='upper left')
    ax.grid(True, alpha=0.3, axis='y')

    # Add totals at the top
    total_original = sum(original_instances)
    total_weighted = sum(smart_weighted_instances)

    ax.text(0.02, 0.98, f'Total normal instances: {total_original:,}',
           transform=ax.transAxes, fontsize=12, fontweight='bold',
           verticalalignment='top', bbox=dict(boxstyle='round', facecolor='lightcoral', alpha=0.7))

    ax.text(0.02, 0.92, f'Total weighted instances: {total_weighted:,}',
           transform=ax.transAxes, fontsize=12, fontweight='bold',
           verticalalignment='top', bbox=dict(boxstyle='round', facecolor='lightgreen', alpha=0.7))

    plt.tight_layout()
    plt.subplots_adjust(top=0.85)
    plt.show()

    # === 📊 PRINT SUMMARY STATISTICS ===
    print("\\n" + "="*80)
    print("📊 DETAILED COMPARISON RESULTS")
    print("="*80)

    print(f"\\n{'Class':<18} {'Original':<10} {'Weighted':<10} {'Ratio':<8} {'Status'}")
    print("-" * 70)

    for cls, orig, weighted in zip(classes, original_instances, smart_weighted_instances):
        ratio = weighted / orig

        if ratio >= 3.0:
            status = "🚀 BOOSTED"
        elif ratio > 1.5:
            status = "📈 IMPROVED"
        elif ratio > 0.9:
            status = "✅ STABLE"
        elif ratio >= 0.8:
            status = "🛡️ PROTECTED"
        else:
            status = "🔧 CAPPED"

        print(f"{cls:<18} {orig:<10,} {weighted:<10,} {ratio:<8.2f} {status}")

    print(f"\\nTotal instances change: {total_weighted/total_original:.2f}x")

    return fig

# === 🚀 EXECUTE THE FUNCTION ===
print("🎯 Creating comparison chart...")
chart = create_comparison_chart()

# === 💾 OPTIONAL: SAVE THE CHART ===
# Uncomment the line below to save the chart
# chart.savefig('/content/class_distribution_comparison.png', dpi=300, bbox_inches='tight')
# print("💾 Chart saved as 'class_distribution_comparison.png'")

print("\\n✅ Chart creation complete!")
```
